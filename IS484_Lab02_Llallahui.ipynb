## **Representación de un problema mediante estados**

#### **Parte A**: Completar la cha del problema para el siguiente caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

| ESTADO | Descripcion |
| --- | --- |
| **Estado inicial** | El repartidor se encuentra en el almacén |
| **Estado objetivo** | El repartidor ha llegado a Zona Norte |
| **Acciones posibles** | El repartidor se puede desplazar a cualquier punto cercano a él |
| **Costo de cada acción** | Consideramos el valor de 1 para cualquier tramo |

#### **Parte B**: Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):

```python
mapa = {
"Almacen": ["Centro", "Terminal"],
"Centro": ["Almacen", "ZonaNorte", "Mercado"],
"Terminal": ["Almacen", "Mercado"],
"Mercado": ["Centro", "Terminal", "ZonaNorte"],
"ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

### **Implementación de BFS**

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """

    cola = [[inicio]]  # cola de caminos, empieza con el estado inicial
    visitados = set()

    while cola:
        camino = cola.pop(0)  # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)

            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)
    
    return None  # no se encontro un camino


resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

### **Vectores y matrices con NumPy**

#### **Parte A**: Representar cuatro observaciones de un dataset simple (por ejemplo, sensores de temperatura y humedad) como un vector cada una, y juntarlas en una matriz:

```python
import numpy as np
obs1 = np.array([22.5, 60]) # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

#### **Parte B**: Acceder a una la y a una columna especíca de la matriz:

```python
print("Primera observacion (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```

### **Producto escalar y distancia euclidiana**

#### **Parte A**: Calcular el producto escalar entre dos observaciones:

```python
producto = np.dot(obs1, obs2)
print("Producto escalar entre obs1 y obs2:", producto)
```

#### **Parte B**: Calcular la distancia euclidiana entre dos observaciones, primero con la fórmula manual y luego con NumPy:

```python
# Formula manual
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)
# Con la funcion de NumPy
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
```
